```python
%%writefile app.py
import streamlit as st
from pypdf import PdfReader
from google import genai

# UI Theme & Page Setup
st.set_page_config(page_title="AI Content Localizer", page_icon="📚", layout="wide")

st.title("📚 AI-Generated Localized Educational Content")
st.subheader("Transform English Textbooks into Local Tamil/Tanglish Context")

# Sidebar for API Key
st.sidebar.header("Settings ⚙️")
api_key = st.sidebar.text_input("Enter your Gemini API Key", type="password")

# Radio button for Language Selection
lang_choice = st.radio(
    "Choose Output Slang/Language:",
    ("Tanglish (WhatsApp chatting style)", "Pure Tamil (Village slang)")
)

st.divider()

# File Uploader
uploaded_file = st.file_uploader("Drop your PDF Textbook here...", type="pdf")

if uploaded_file and api_key:
    if st.button("Localize Content 🔥", type="primary"):
        with st.spinner("Wait machi, AI is working on it... ⏳"):
            try:
                # 1. Extract Text
                reader = PdfReader(uploaded_file)
                text_data = ""
                for page in reader.pages:
                    text = page.extract_text()
                    if text:
                        text_data += text + "\n"

                # 2. Setup AI
                client = genai.Client(api_key=api_key)
                if "Tanglish" in lang_choice:
                    prompt_lang = "Tanglish (Tamil words written entirely in English alphabets, like how friends chat on WhatsApp. Do NOT use Tamil letters)."
                else:
                    prompt_lang = "Pure Tamil script using rural village slang and relatable local examples."

                prompt_text = f"""
                You are an expert Educational Content Localizer working for rural students in Tamil Nadu.
                Take the following extracted textbook content and rewrite it in an engaging, localized way.
                Explain the concepts using relatable real-world examples (like farming, local shops, or local sports).

                CRITICAL INSTRUCTION: You MUST generate the output EXACTLY in this language/slang: {prompt_lang}.

                Here is the extracted content:
                {text_data[:1000]}
                """

                # 3. Call AI
                response = client.models.generate_content(
                    model="gemini-2.5-flash",
                    contents=prompt_text,
                )

                st.success("Success! Check the localized output below 🎉")

                # Final Output Display
                st.write("### 🎓 Translated Local Context:")
                st.info(response.text)

            except Exception as e:
                st.error(f"Oops! Oru error vandhuduchu: {e}")
elif not api_key:
    st.warning("👈 Left side-la un API key-a podu machi!")
```

```python
%%writefile app.py
import streamlit as st
from pypdf import PdfReader
from google import genai

# UI Theme & Page Setup
st.set_page_config(page_title="AI Localized Tutor", page_icon="🎓", layout="wide")
st.title("🎓 AI Conversational Localized Tutor")
st.subheader("Chat with your textbook in local Tamil/Tanglish slang!")

# --- Sidebar (Settings & PDF Upload) ---
st.sidebar.header("Settings ⚙️")
api_key = st.sidebar.text_input("Enter your Gemini API Key", type="password")

lang_choice = st.sidebar.radio(
    "Choose AI Chat Language:",
    ("Tanglish (WhatsApp chatting style)", "Pure Tamil (Village slang)")
)
st.sidebar.divider()

uploaded_file = st.sidebar.file_uploader("1. Drop your PDF Textbook here", type="pdf")

# --- Session State (To remember chat history & PDF text) ---
if "messages" not in st.session_state:
    st.session_state.messages = []

if "pdf_text" not in st.session_state:
    st.session_state.pdf_text = ""

# --- PDF Extraction ---
if uploaded_file and not st.session_state.pdf_text:
    with st.spinner("Extracting textbook context... ⏳"):
        reader = PdfReader(uploaded_file)
        text_data = ""
        for page in reader.pages:
            text = page.extract_text()
            if text:
                text_data += text + "\n"
        st.session_state.pdf_text = text_data
        st.sidebar.success("✅ PDF Loaded into AI Memory!")

# --- Chat Interface ---
# Display previous chat messages
for msg in st.session_state.messages:
    with st.chat_message(msg["role"]):
        st.markdown(msg["content"])

# User Chat Input Box
if user_question := st.chat_input("Ask a doubt from the PDF (e.g., 'Momentum na enna?')..."):
    if not api_key:
        st.warning("👈 Please enter your API Key in the sidebar first!")
    elif not st.session_state.pdf_text:
        st.warning("👈 Please upload a PDF textbook first!")
    else:
        # 1. Display user question
        st.session_state.messages.append({"role": "user", "content": user_question})
        with st.chat_message("user"):
            st.markdown(user_question)

        # 2. AI Response Generation
        with st.chat_message("assistant"):
            with st.spinner("AI is thinking... 🤔"):
                try:
                    client = genai.Client(api_key=api_key)

                    if "Tanglish" in lang_choice:
                        prompt_lang = "Tanglish (Tamil words written entirely in English alphabets, like WhatsApp chat. Do NOT use Tamil letters)."
                    else:
                        prompt_lang = "Pure Tamil script using rural village slang and relatable local examples."

                    # System Prompt + Context (RAG)
                    full_prompt = f"""
                    You are an expert, friendly Conversational AI Tutor for rural students in Tamil Nadu.
                    Answer the student's question based ONLY on the provided textbook content.

                    CRITICAL INSTRUCTIONS FOR FORMAT & LOCALIZATION:
                    1. STRICT LOCAL VOCABULARY: {prompt_lang}. Do NOT mix English words for common things (e.g., strictly use 'Kenaru' instead of 'well', 'Kaichal' instead of 'fever', 'Vivasayam' instead of 'farming'). Keep the vocabulary 100% authentic local.
                    2. NEAT ORGANIZATION: Do NOT write messy, long paragraphs. You MUST use clear bullet points, very short sentences, bold text for highlights, and emojis for readability. Keep the structure highly organized.
                    3. FRIENDLY TONE: Talk like a close friend or friendly senior teaching a subject ("Machi", "Da", "Puriyudha"). Keep the conversation engaging but to the point.
                    4. CULTURAL ANALOGIES: Use relatable real-world village analogies (Panchayat, local shops, farming, local sports) and explain them crisply without dragging.

                    Textbook Context:
                    {st.session_state.pdf_text[:3000]}

                    Student's Question: {user_question}
                    """

                    response  = client.models.generate_content(
                        model="gemini-3.6-flash",
                        contents=full_prompt,
                    )

                    st.markdown(response.text)
                    # Add AI response to history
                    st.session_state.messages.append({"role": "assistant", "content": response.text})

                except Exception as e:
                    st.error(f"Oops! Server error: {e}")
```
